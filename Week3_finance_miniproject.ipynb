````python
print("=== Budget Limit Checker ===")

# Get user's budget and expense
monthly_budget = float(input("Enter your monthly budget: $"))
expense = float(input("Enter expense amount: $"))

# Decision logic here - you write this part!
# If expense > 50% of budget: "Major purchase - think carefully"
# If expense > 25% of budget: "Significant expense"
# If expense > 10% of budget: "Moderate expense"
# Otherwise: "Minor expense - within budget"
````

````python
# Expense Classifier
amount = float(input("Enter expense amount: $"))

if amount >= 250:
    category = "Major Expense"
    message = "Consider if its really a need or a want!"
elif amount >= 150:
    category = "Moderate Expense"
    message = "Make sure this is within your budget."
elif amount >= 50:
    category = "Minor Expense"
    message = "Reasonable - keep track of minor expenses!"
else:
    category = "Small Purchase"
    message = "Small bills can add up overtime!"

print(f"\nCategory: {category}")
print(f"Message:  {message}")
````

### Savings Goal Progress
Use multiple conditions to track savings progress:

````python
print("=== Savings Goal Tracker ===")

ggoal = float(input("Savings goal: $"))
saved = float(input("Current savings: $"))

percentage = (saved / ggoal) * 100
remaining = ggoal - saved

if saved >= ggoal:
    status = "GOAL ACHIEVED. Consider a new goal!"
    advice = f"Amazing! You've hit your ${ggoal:.2f} goal."
elif percentage >= 75:
    status = "ALMOST THERE! You're doing great."
    advice = f"Just ${remaining:.2f} to go — don't stop now!"
elif percentage >= 50:
    status = "HALFWAY THERE"
    advice = f"Great progress! ${remaining:.2f} still needed."
elif percentage >= 25:
    status = "MAKING PROGRESS"
    advice = f"You've saved {percentage:.1f}%. Keep it up!"
else:
    status = "JUST GETTING STARTED"
    advice = f"Every dollar saved matters. ${remaining:.2f} to reach your goal."

print(f"\nProgress:  {percentage:.1f}% complete")
print(f"Status:    {status}")
print(f"Advice:    {advice}")

# Your task: Calculate percentage and give encouraging feedback
# 100%+: "Goal achieved! Consider a new goal."
# 75-99%: "Almost there! You're doing great."
# 50-74%: "Halfway there - keep it up!"
# 25-49%: "Good progress, stay consistent"
# Under 25%: "Just getting started - you've got this!"
````

## Part 3: Smart Purchase Advisor (20 minutes)

### The Challenge
Build a program that helps users make smart purchase decisions by considering multiple factors.

### Requirements

Your purchase advisor should:

1. Get purchase amount and user's available money
2. Check if they can afford it (basic affordability)
3. Give different advice based on purchase size
4. Consider their financial situation

### Starter Framework

````python
# Smart Purchase Advisor
print("=== Smart Purchase Advisor ===\n")

# --- Inputs ---
available = float(input("Available money: $"))
purchase = float(input("Purchase amount: $"))
monthly_income = float(input("What's your monthly income? $"))
emergency_fund = float(input("Enter your emergency fund amount (for comparison): $"))
category = input("Category (necessity/luxury): ").strip().lower()

# --- Calculations ---
# Handle potential ZeroDivisionError if available is 0
if available == 0:
    percentage = float('inf') if purchase > 0 else 0.0
else:
    percentage = (purchase / available) * 100
remaining = available - purchase
can_afford = purchase <= available
shortfall = purchase - available
would_drain_emergency = remaining < emergency_fund

print("\n" + "=" * 40)

# --- Core Affordability Check ---
if not can_afford:
    print(f"CANNOT AFFORD")
    print(f"You need ${shortfall:.2f} more before buying this.")

elif percentage >= 75:
    print(f"VERY EXPENSIVE — {percentage:.1f}% of your money")
    print(f"This is a want rather than a need")


elif percentage >= 40:
    print(f"SIGNIFICANT PURCHASE — {percentage:.1f}% of your money")
    print(f"Consider if this purchuse gonna benefit you and earn you back money")

elif percentage >= 15:
    print(f"REASONABLE PURCHASE — {percentage:.1f}% of your money")
    print(f"This fits your budget well.")

else:
    print(f"MINOR EXPENSE — {percentage:.1f}% of your money")
    print(f"Go for it!")

# --- Category-Based Advice ---
if category == "luxury" and percentage >= 40:
    print(f"Luxury tip: Since this is a want (not a need),")
    print(f"wait until you gathered all the money, rethink, act, and bare consequences")

elif category == "necessity" and not can_afford:
    print(f"Necessity tip: Since this is essential,")
    print(f":Look for payment plans or lower-cost alternatives.")

# --- Remaining Balance ---
if can_afford:
    print(f"\n💰 Remaining after purchase: ${remaining:.2f}")

print("=" * 40)

# Your task: Build the decision logic
# Consider:
# - Can they afford it at all?
# - Is it a large percentage of their available money?
# - Is it reasonable compared to their income?
# - What advice should you give?

# Suggestion: Use multiple if/elif conditions to give personalised advice
````
