This notebook walks through a structured problem-solving methodology to build the first component of a semester-long Personal Finance Tracker: the **user profile setup**. We will collect basic user preferences and display a personalised welcome message.

## Step 1 – State the Problem

**Goal:**  
Build a program that creates a user profile for a personal finance tracker by collecting a few initial settings from the user, then displays a personalised welcome message summarising their preferences.

**Data to Gather:**
- The user's **name** — so the app can greet them personally.
- Their preferred **theme colour** — so the interface can be customised (e.g., dark, blue, green).
- Their **primary currency** — so all monetary amounts display correctly (e.g., AUD, USD, EUR).
- Their **main savings goal** — so the app can keep them motivated (e.g., Holiday, New Car).

**Constraints:**  
The solution must only use `input()`, `print()`, basic variables, and string concatenation — no advanced Python features.

## Step 2 – Understand the Problem

Before writing any code, let's make sure we understand what the program needs to do:

1. **Inputs:** Four pieces of information collected interactively from the user via `input()`.
2. **Processing:** Combine those four values into a single, readable summary string.
3. **Output:** Print that summary string as a personalised welcome message.

There is no calculation or decision-making required at this stage — it is purely about collecting, storing, and displaying information.

## Step 4 – Develop an Algorithm (Pseudocode)

```
1. Start the program.
2. Prompt the user to enter their name → store in user_name.
3. Prompt the user to enter their preferred theme colour → store in theme_color.
4. Prompt the user to enter their primary currency → store in primary_currency.
5. Prompt the user to enter their main savings goal → store in main_savings_goal.
6. Build a welcome message by combining all four variables into one string.
7. Print the welcome message.
8. End the program.
```

## Step 5 – Write the Code

Now we translate the pseudocode directly into Python.

````python
# Import datetime so we can timestamp the profile creation
from datetime import datetime

# ── Step 1: Collect user profile information ──────────────────────────────────
# Each input() call pauses the program and waits for the user to type a response.
# The response is stored as a string in a descriptively named variable.

user_name         = input("What is your name? ")
theme_color       = input("What colour would you like for your app theme? (e.g., blue, green, dark) ")
primary_currency  = input("What is your primary currency? (e.g., AUD, USD, EUR) ")
main_savings_goal = input("What is one big savings goal you have? (e.g., New Car, Holiday) ")

# ── Step 2: Build the personalised welcome message ────────────────────────────
# String concatenation joins the variables with fixed text to form one sentence.
# The backslash (\) allows the long string to continue on the next line — 
# this is purely for readability in the notebook; Python treats it as one string.

summary = ("Welcome to your Personal Finance Tracker, " + user_name + "! "
           "Your profile is ready. "
           "The app will use a " + theme_color + " theme and all amounts will be in "
           + primary_currency + ". "
           "Let's start working towards your goal of saving for a "
           + main_savings_goal + "!")

# ── Step 3: Display the result ────────────────────────────────────────────────
print()
print(summary)
````
