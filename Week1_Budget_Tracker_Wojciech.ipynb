## Part 3: Building the Budget Tracker

Use AI to help you write a program that:
- Allows users to input their monthly income
- Allows users to input up to 3 expenses  
- Calculates remaining budget
- Shows a warning if budget is negative

### 💡 Pro Tip: Be specific about requirements!

````python
# TASK: Write your detailed prompt for the budget tracker
"""
My Budget Tracker Prompt:
[Write your detailed prompt here]
"""
````

### Paste and Test Your AI-Generated Budget Tracker

````python
# AI-GENERATED BUDGET TRACKER CODE
# # Budget Tracker Program

# Get monthly income
income = float(input("Enter your monthly income: $"))

# Get up to 3 expenses
print("\nEnter up to 3 expenses:")
expense1 = float(input("Expense 1: $"))
expense2 = float(input("Expense 2: $"))
expense3 = float(input("Expense 3: $"))

# Calculate total expenses
total_expenses = expense1 + expense2 + expense3

# Calculate remaining budget
remaining_budget = income - total_expenses

# Display results
print("\n" + "="*40)
print("📊 BUDGET SUMMARY")
print("="*40)
print(f"Monthly Income: ${income:.2f}")
print(f"Total Expenses: ${total_expenses:.2f}")
print(f"Remaining Budget: ${remaining_budget:.2f}")

# Show warning if budget is negative
if remaining_budget < 0:
    print("\n⚠️ WARNING: You are over budget!")
    print(f"You are ${abs(remaining_budget):.2f} over your budget.")
else:
    print("\n✅ Great! You're within your budget.")

# ```
# **Features of this program:**
# - Uses `float()` to handle decimal numbers for money
# - Calculates total expenses automatically
# - Uses `.2f` formatting to show exactly 2 decimal places (standard for money)
# - Displays a warning with ⚠️ emoji if budget goes negative
# - Shows how much over budget if negative
# - Shows a positive message if within budget

# **Example output:**
# ```
# Enter your monthly income: $3000

# Enter up to 3 expenses:
# Expense 1: $1200
# Expense 2: $500
# Expense 3: $800

# ========================================
# 📊 BUDGET SUMMARY
# ========================================
# Monthly Income: $3000.00
# Total Expenses: $2500.00
# Remaining Budget: $500.00

# ✅ Great! You're within your budget.
# ```
````

## Part 4: Enhancement Challenge

Choose ONE feature to add using AI assistance:
- 🏷️ Expense categories (food, transport, entertainment)
- 📊 Show percentage of income spent
- ✅ Input validation (no negative numbers)
- 📝 Detailed budget summary with advice

````python
enhancement_choice = "I chose to add: Expense categories (food, transport, entertainment)"
enhancement_prompt = """
Write a Python program for a budget tracker that:
- Asks the user for their monthly income.
- Allows the user to input multiple expenses, each with a category (e.g., 'Food', 'Transport', 'Entertainment') and an amount. The user should be able to type 'done' to stop adding expenses.
- Calculates the total expenses for each category.
- Calculates the remaining budget.
- Displays a detailed summary, including:
  - Monthly Income
  - Total expenses per category
  - Total overall expenses
  - Remaining budget
- Shows a warning if the remaining budget is negative, indicating how much over budget the user is.
- Includes basic input validation to ensure income and expense amounts are positive numbers.
"""

print(f"🚀 ENHANCEMENT: {enhancement_choice}")
````

````python
# ENHANCED BUDGET TRACKER
# Function to get a valid positive float input
def get_positive_float_input(prompt):
    while True:
        try:
            value = float(input(prompt))
            if value < 0:
                print("Amount cannot be negative. Please enter a positive number.")
            else:
                return value
        except ValueError:
            print("Invalid input. Please enter a number.")

# Get monthly income
income = get_positive_float_input("Enter your monthly income: $")

# Dictionary to store expenses by category
expenses = {}

print("\nEnter your expenses. Type 'done' for category when finished.")
while True:
    category = input("Expense category (e.g., Food, Transport, Entertainment, or 'done'): ").strip()
    if category.lower() == 'done':
        break
    if not category:
        print("Category cannot be empty. Please enter a category or 'done'.")
        continue

    amount = get_positive_float_input(f"Amount for {category}: $")
    expenses[category] = expenses.get(category, 0) + amount

# Calculate total overall expenses
total_overall_expenses = sum(expenses.values())

# Calculate remaining budget
remaining_budget = income - total_overall_expenses

# Display results
print("\n" + "="*40)
print("📊 DETAILED BUDGET SUMMARY")
print("="*40)
print(f"Monthly Income: ${income:.2f}")
print("\n--- Expenses by Category ---")
if expenses:
    for category, amount in expenses.items():
        print(f"  {category}: ${amount:.2f}")
else:
    print("  No expenses entered.")
print(f"\nTotal Overall Expenses: ${total_overall_expenses:.2f}")
print(f"Remaining Budget: ${remaining_budget:.2f}")

# Show warning if budget is negative
if remaining_budget < 0:
    print("\n⚠️ WARNING: You are over budget!")
    print(f"You are ${abs(remaining_budget):.2f} over your budget.")
else:
    print("\n✅ Great! You're within your budget.")
````
